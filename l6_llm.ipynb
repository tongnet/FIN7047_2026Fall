```python
import PyPDF2
from openai import OpenAI
from getpass import getpass

# Enter the API key securely
client = OpenAI(
    api_key=getpass("Enter your OpenAI API key: ")
)

# Extract text from the PDF
def extract_text_from_pdf(file_path):
    text = ""

    with open(file_path, "rb") as file:
        reader = PyPDF2.PdfReader(file)

        for page in reader.pages:
            text += (page.extract_text() or "") + "\n"

    return text

pdf_text = extract_text_from_pdf(file_path)

messages = [
    {
        "role": "system",
        "content": (
            "You are a helpful assistant specializing in "
            "cryptocurrency and financial systems."
        )
    },
    {
        "role": "user",
        "content": f"""
Analyze the following excerpt from Nakamoto's 2008 paper:

{pdf_text[:1500]}

What are the key concepts and the main problem addressed in the paper?
"""
    }
]

response = client.chat.completions.create(
    model="gpt-4-turbo",
    messages=messages,
    max_tokens=500,
    temperature=0.7
)

print("GPT Analysis of the Article:")
print(response.choices[0].message.content)
```

```python
# Break the extracted PDF text into chunks
chunk_size = 1500

text_chunks = [
    pdf_text[i:i + chunk_size]
    for i in range(0, len(pdf_text), chunk_size)
]

# Analyze one text chunk
def gpt_topic_analysis(chunk):
    response = client.chat.completions.create(
        model="gpt-4-turbo",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an expert in cryptocurrencies "
                    "and financial systems."
                )
            },
            {
                "role": "user",
                "content": (
                    "Analyze the following text and identify its "
                    "main topics and subtopics:\n\n" + chunk
                )
            }
        ],
        max_tokens=200,
        temperature=0.7
    )

    return response.choices[0].message.content.strip()

# Analyze every chunk
topic_results = []

for i, chunk in enumerate(text_chunks):
    print(f"Analyzing chunk {i + 1} of {len(text_chunks)}...")

    topic_summary = gpt_topic_analysis(chunk)

    topic_results.append(
        f"Chunk {i + 1} Analysis:\n{topic_summary}"
    )

# Print all results
print("\n=== Combined Topic Analysis ===\n")

for result in topic_results:
    print(result)
    print()



```
